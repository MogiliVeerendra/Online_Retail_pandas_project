```python
import pandas as pd
import numpy as np
```

```python
file_path = "../data/interim/featured_online_retail.xlsx"
sales = pd.read_excel(file_path)
```

```python
sales.head()
```

```python
sales.select_dtypes(include=["object","str"])
```

```python
sales = pd.get_dummies(
    sales,
    columns= ["Country"],
    dtype= int
)
sales
```

```python
sales["DayName"]
```

```python
sales = pd.get_dummies(
    sales,
    columns=["DayName"],
    dtype= int
)
sales
```

```python
sales = sales.drop(columns=["DayName_Friday","DayName_Monday","DayName_Sunday","DayName_Thursday","DayName_Tuesday","DayName_Wednesday"])
```

```python
sales["Day_sin"] = np.sin(2 * np.pi * sales["Dayofweek"] / 7)
sales["Day_cos"] = np.cos(2 * np.pi * sales["Dayofweek"] / 7)

sales["Month_sin"] = np.sin(2 * np.pi * sales["Month"] / 12)
sales["Month_cos"] = np.cos(2 * np.pi * sales["Month"] / 12)

```

```python
sales = sales.drop(columns = ["MonthName"])
```

```python
sales = sales.drop(columns = ["Description"])
```

```python
sales.to_excel("../data/interim/encoded_online_retail.xlsx")
```
