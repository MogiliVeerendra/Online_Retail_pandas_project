```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "../data/raw/Online_Retail.xlsx"
sales = pd.read_excel(file_path)
```

```python
sales["Quantity"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(sales["Quantity"],bins= 20)
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.title("Distribution of Quantity")
plt.show()
```

```python
sales["UnitPrice"].describe()
```

```python
fig,ax = plt.subplots()

ax.set_title("Distribution of Unit Price")
ax.hist(sales["UnitPrice"], bins = 50)
ax.set_xlabel("unit price")
ax.set_ylabel("Frequency")
plt.show()
```

```python
fig,ax = plt.subplots()

ax.boxplot(sales["Quantity"])
ax.set_ylabel("Quantity")
plt.show()
```

```python
fig,ax = plt.subplots()

ax.boxplot(sales["UnitPrice"])
ax.set_ylabel("unit price")

plt.show()
```

```python
fig, ax = plt.subplots()
country_counts = sales["Country"].value_counts().head()
ax.bar(country_counts.index, country_counts.values)
plt.show()
```

```python
fig,ax = plt.subplots()

top_products = sales["Description"].value_counts().head(10)
ax.barh(top_products.index,top_products.values)

plt.show()
```

```python
fig,ax = plt.subplots()

ax.scatter(sales["Quantity"],sales["UnitPrice"],alpha= 0.3)

plt.show()
```
