```python
import pandas as pd

```

```python
file_path = "../data/raw/online_Retail.xlsx"
sales = pd.read_excel(file_path)
sales
```

```python
quality_report = pd.DataFrame({
    "columns": sales.columns,
    "data types": sales.dtypes.astype(str).values,
    "missing count": sales.isna().sum(),
    "missing percentage": (sales.isna().mean() * 100).values,
    "duplicated": sales.duplicated().sum(),
    "unique count": sales.nunique().values
})
quality_report
```
