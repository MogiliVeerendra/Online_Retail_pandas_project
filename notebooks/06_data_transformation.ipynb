```python
import pandas as pd
```

```python
file_path = "../data/interim/cleaned_online_retail.xlsx"
sales = pd.read_excel(file_path)
sales
```

```python
sales.head()
```

```python
sales["Year"] = sales["InvoiceDate"].dt.year
sales["Month"] = sales["InvoiceDate"].dt.month
sales["Day"] = sales["InvoiceDate"].dt.day
sales["Hour"] = sales["InvoiceDate"].dt.hour
sales["Dayofweek"] = sales["InvoiceDate"].dt.dayofweek
sales["DayName" ] = sales["InvoiceDate"].dt.day_name()
sales["MonthName"] = sales["InvoiceDate"].dt.month_name()
```

```python
sales["Description"].head(10)
```

```python
sales["Description"].str.strip()
sales["Description"].str.upper()
```

```python
sales["Country"].str.strip()
sales["Country"].str.upper()
```

```python
sales["TotalAmount"] = sales["Quantity"] * sales["UnitPrice"]
```

```python
print(sales.isna().sum())
print(sales.dtypes)
print(sales.shape)
```

```python
sales = sales.drop(columns = ["TotalAmount"])
```

```python
sales.to_excel("../data/interim/tranformaed_online_retail.xlsx", index = False)
```
